```python
import numpy as np
```

## Defining the basis states
We start by defining numpy arrays for the basis states $\ket{0}$ and $\ket{1}$. 

( *NOTE: All matrices need to be in the `complex64` datatype* )

```python
k0 = np.array([[1],[0]]).astype(np.complex64)
k1 = np.array([[0],[1]]).astype(np.complex64)
```

```python
print(k0, '\n','-'*20,'\n' ,k1)
```

## Creating State from basis
As we know we can create states by writing them as superpositions or linear combinations of the basis states. Let's create the following state :
$$
    \ket{\psi} = \frac{1}{\sqrt{3}}\ket{0} + \sqrt{\frac{2}{3}}\ket{1}
$$

```python
psi = np.sqrt(1/3)*k0 + np.sqrt(2/3)*k1
```

```python
print(psi)
```

## Define Gate Matrices
We now define matrices for the 4 main single qubit gates

```python
X = np.array([[0,1],[1,0]]).astype(np.complex64)
Y = np.array([[0+0j,0+1j],[0-1j,0]]).astype(np.complex64)
Z = np.array([[1,0],[0,-1]]).astype(np.complex64)
H = (1/np.sqrt(2))*np.array([[1,1],[1,-1]]).astype(np.complex64)
```

## Applying Gate
As we know we can apply gates via matrix multiplications. So we simply multiply the gates to the state. Lets apply the following computation first:
$$
    \ket{\psi_1} = X\ket{\psi}
